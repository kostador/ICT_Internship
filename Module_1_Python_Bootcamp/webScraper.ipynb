```python
class Menu:
    def __init__(self):
        self.name = "menu name unspecified"
        self.menuOptions = 3
        self.backOption = "backmenu unspecified"
    def scrape(self):
        pass
    def __str__(self) -> str:
        return self.name
    def backMenu(self):
        print("backmenu ....")
    def start(self):
        returnFlag = False
        while not returnFlag:
            print(self.name)
            print(f"{0}. {self.backOption}")
            for i in range(1,self.menuOptions-1):
                print(f"{i}. {i}")
            validInputs = set(range(self.menuOptions))
            while True:
                userInput = input("Enter an option by index: ")
                if int(userInput) not in validInputs:
                    print("Invalid input")
                    continue
                userInput = int(userInput)
                # backmenu decision
                if userInput == 0:
                    returnFlag = True
                    self.backMenu()
                    break
                else:
                    print(f"Selecting menu {userInput} ....")
                    break
```

```python
class Menu:
    def __init__(self,title = "menu title unspecified", parent = None, menuOptions = {}): #parent is the parent menu
        self.title = title
        self.parent = parent
        self.menuOptions = menuOptions
    def display(self):
        print(f"\n### {self.title} ###")

        for number, (description,_) in self.menuOptions.items():
            print(f"{number}. {description}")

        if self.parent:
            print("0. Back")
        else:
            print("0. Quit")
    def run(self):
        current_menu = self
        while current_menu is not None:
            current_menu.display()
            choice = input("Select an option: ").strip()

            if choice == "0":
                current_menu = current_menu.parent
            elif choice in current_menu.menuOptions:
                description, action = current_menu.menuOptions[choice]

                if isinstance(action, Menu): # if selected action is a menu, we go to the menu. Else it must be an action so execute it
                    current_menu = action
                else:
                    action()

            else:
                print("invalid input")

class TestMenu(Menu):
    def __init__(self, parent = None):
        super().__init__(title = "Test menu",
                         parent = parent,
                         menuOptions = {
                             "1": ("Test print action", self.testPrintAction)
                         })
    def testPrintAction(self):
        print("testPrintAction() was called")

class BooksMenu(Menu):
    def __init__(self, parent = None):
        super().__init__(title = "books.toscrape.com scraper",parent = parent,menuOptions = {})


class MobileMenu(Menu):
    def __init__(self, parent = None):
        super().__init__(title = "mobile.bg scraper", parent = parent, menuOptions = {})


class MainMenu(Menu):
    def __init__(self):
        super().__init__("Main Menu")
        self.testMenu = TestMenu(parent=self)
        self.booksMenu = BooksMenu(parent = self)
        self.mobileMenu = MobileMenu(parent = self)

        self.menuOptions = {
            "1": ("testing menu", self.testMenu),
            "2": ("books.toscrape.com", self.booksMenu),
            "3": ("mobile.bg", self.mobileMenu),
        }
```

```python
menu1 = MainMenu()
menu1.run()
```
