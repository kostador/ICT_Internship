```python
from IPython.display import clear_output
```

```python

validInputs = {'1','2','3','4','5','6','7','8','9'}

def checkBoard(board, playerSymbol):

    # check if a player won

    # Не намерих начин да го абстрактирам това без да се ползват матрици от numpy например
    winMark = playerSymbol*3
    if board[0]+board[1]+board[2] == winMark or board[3]+board[4]+board[5] == winMark or board[6]+board[7]+board[8] == winMark or board[0]+board[3]+board[6] == winMark or board[1]+board[4]+board[7] == winMark or board[2]+board[5]+board[8] == winMark or board[0]+board[4]+board[8] == winMark or board[2]+board[4]+board[6] == winMark:
        return playerSymbol
    # tie if the board is full
    # board is full when there are no " "
    elif " " not in board:
        return "Tie"
    # otherwise continue
    else:
        return "ContinueGame"


def displayBoard(board):
    print(f'{board[6]}|{board[7]}|{board[8]}\n{board[3]}|{board[4]}|{board[5]}\n{board[0]}|{board[1]}|{board[2]}\n')


def playerMakeTurn(board, playerSymbol):
    while True:

        playerInput = input("Player " + playerSymbol + "  input: ")

        # check if the input is a valid
        if playerInput not in validInputs:
            print("Invalid input, try an integer from 1 to 9")
            continue
        playerInput = int(playerInput)

        # check if the square is taken
        if board[playerInput - 1] != " ":
            print("Square already taken, try again")
            continue
        else: # else perform move
            board[playerInput - 1] = playerSymbol
            break
```

```python
quitGame = False
while not quitGame:
    #initialize game
    board = [" "]*9
    boardState = "ContinueGame"
    playerOneSymbol = "X"
    playerTwoSymbol = "O"
    clear_output()
    displayBoard(board)

    #start repeating turns
    while True:
        # player X's turn
        print("Player " + playerOneSymbol + "'s turn: ")
        playerMakeTurn(board, playerOneSymbol)
        clear_output()
        displayBoard(board)
        boardState = checkBoard(board, playerOneSymbol)
        if boardState != "ContinueGame":
            break

        # player O's turn
        print("Player " + playerTwoSymbol + "'s turn: ")
        playerMakeTurn(board, playerTwoSymbol)
        clear_output()
        displayBoard(board)
        boardState = checkBoard(board, playerTwoSymbol)
        if boardState != "ContinueGame":
            break

    # check for winner
    if boardState != "ContinueGame":
        if boardState == "Tie":
            print("The game is a tie!")
        else:
            print("Player " + boardState + " wins!")

    # ask for a new game
    pInput = ""
    while pInput != "N" and pInput != "Y":
        pInput = input("Play again? (Y/N): ")
        if pInput != "Y" and pInput != "N":
            print("Invalid input")
            continue

    if pInput == "Y":
        quitGame = False
    else:
        quitGame = True
        print("Thank you for playing!")




```
